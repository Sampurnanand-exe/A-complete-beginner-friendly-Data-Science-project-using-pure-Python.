## Task 2: Clean the Data

Clean the dataset by following these rules:

1. **Remove users with missing names.**
2. **Remove duplicate friend entries.**
3. **Remove inactive users** with no friends and no liked pages.
4. **Remove duplicate pages** based on their IDs.

### Goal

Create a clean and well-structured dataset using **pure Python**.

````python
def clean_data(data):
    
    # Remove users with missing names
    data['users'] = [user for user in data['users'] if user['name'].strip()]

    #  Remove duplicate friend entries
    for user in data['users']:
        user['friends'] = list(set(user['friends']))

    # Remove inactive users
    data['users'] = [user for user in data['users'] if user['friends'] or user['liked_pages']]

    # Remove duplocate pages
    unique_pages = {}
    for page in data['pages']:
        unique_pages[page['id']] = page
    data['pages'] = list(unique_pages.values())

    return data

# Load , Clean and display the cleaned data

file = json.load(open("../Data/codeBook.json"))
data = clean_data(file)

json.dump(data, open("../Data/Cleaned_Data.json", "w"), indent=4)
print("Data has been cleaned successfully!")
````
